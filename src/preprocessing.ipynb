```python
import pandas as pd
from sklearn.preprocessing import StandardScaler


def load_data(path: str) -> pd.DataFrame:
    """
    Load raw soil dataset
    """
    df = pd.read_csv(path)
    return df


def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Handle missing values and basic cleaning
    - Numeric columns: fill missing values with mean
    - Categorical columns: fill missing values with "Unknown"
    """
    # Fill missing numerical values with mean
    df = df.fillna(df.mean(numeric_only=True))
    # Fill remaining missing values (categorical) with "Unknown"
    df = df.fillna("Unknown")
    return df


def split_features_target(df: pd.DataFrame, target_col: str):
    """
    Split dataset into features and target
    """
    X = df.drop(columns=[target_col])
    y = df[target_col]
    return X, y


def scale_features(X: pd.DataFrame):
    """
    Scale numerical features only; keep categorical columns as-is.
    Returns a DataFrame with same columns, numeric ones scaled.
    """
    # Identify numeric columns
    num_cols = X.select_dtypes(include=["number"]).columns
    
    scaler = StandardScaler()
    X_scaled_array = scaler.fit_transform(X[num_cols])
    
    # Put scaled values back into a copy of X
    X_scaled = X.copy()
    X_scaled[num_cols] = X_scaled_array
    
    return X_scaled, scaler


def preprocess_pipeline(path: str, target_col: str):
    """
    Complete preprocessing pipeline
    """
    df = load_data(path)
    df = clean_data(df)
    X, y = split_features_target(df, target_col)
    X_scaled, scaler = scale_features(X)

    return X_scaled, y, scaler
```

```python
# Set the correct path (relative to project root, not src/)
DATA_PATH = "../data/data/raw/data_core.csv"
TARGET_COL = "Fertilizer Name"

# Run the preprocessing pipeline
X_processed, y, scaler = preprocess_pipeline(DATA_PATH, TARGET_COL)

print("Preprocessing completed!")
print(f"Features shape: {X_processed.shape}")
print(f"Target length: {len(y)}")
print(f"\nFirst few rows of processed features:")
X_processed.head()
```
